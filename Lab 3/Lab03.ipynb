## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
# T3: Clean the data (six types, in order)
# (a) Missing values
print("T3a: missing values per column (before)")
print(df.isna().sum())
rows_before = len(df)
df = df.dropna(subset=["Brand", "Model"])
for col in ["Registered City", "Fuel", "Transaction Type"]:
    df[col] = df[col].fillna("Unknown")
print("T3a: rows removed =", rows_before - len(df), "| rows left =", len(df))

# T3b: Formatting
df["Registered City"] = df["Registered City"].str.strip().str.title()
df["Price"] = df["Price"].str.replace("Rs", "", regex=False).str.replace(",", "", regex=False).str.strip()
print("T3b: sample prices after cleaning text:", df["Price"].head(3).tolist())

# T3c: Data types
df["Price"] = df["Price"].astype("int64")
df["Year"] = df["Year"].astype("Int64")
print("T3c: Price dtype =", df["Price"].dtype, "| Year dtype =", df["Year"].dtype)

# T3d: Inconsistent categories
print("T3d: different cities before =", df["Registered City"].nunique())
df["Registered City"] = df["Registered City"].replace({"Sukkar": "Sukkur"})
df["Model"] = df["Model"].replace({"Corrolla Altis": "Corolla Altis"})
print("T3d: different cities after =", df["Registered City"].nunique())

# T3e: Duplicates
print("T3e: exact duplicate ads =", df.duplicated().sum())
df = df.drop_duplicates()
print("T3e: rows after dropping duplicates =", len(df))

# T3f: Invalid values
bad_km = (df["KMs Driven"] < 100) | (df["KMs Driven"] > 1_000_000) | df["KMs Driven"].isin([12345, 123456])
print("T3f: KMs Driven set to NaN =", bad_km.sum())
df.loc[bad_km, "KMs Driven"] = np.nan

fake_price = df["Price"].isin([12_345_678, 87_654_321])
cheap_brand_too_dear = df["Brand"].isin(["Suzuki", "Daihatsu"]) & (df["Price"] > 5_000_000)
print("T3f: ads dropped for fake prices =", fake_price.sum(), "| Suzuki/Daihatsu above 5M =", cheap_brand_too_dear.sum())
df = df[~(fake_price | cheap_brand_too_dear)]

# T3g: Self-check
print("T3g: final shape =", df.shape)
print("T3g: matches (20484, 9)?", df.shape == (20484, 9))
```

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```python
# T4: Analyse
df["Price_lakh"] = df["Price"] / 100_000
df["Age"] = 2020 - df["Year"]
print("T4a/b: new columns added")
print(df[["Price", "Price_lakh", "Year", "Age"]].head())
print("T4c: top 5 brands")
print(df["Brand"].value_counts().head(5))
print("T4d: top 5 cities by ads")
print(df["Registered City"].value_counts().head(5))
print("T4e: median Price_lakh per fuel type")
print(df.groupby("Fuel")["Price_lakh"].median().sort_values(ascending=False))

gli = df[(df["Model"] == "Corolla GLI") & (df["Year"].between(2010, 2018))]
gli_med = gli.groupby("Year")["Price_lakh"].median()
print("T4f: median Corolla GLI price (lakh) by year")
print(gli_med)
print("T4f: change vs previous year")
print(gli_med.diff())
print("T4f: average gain per year of newer model (2010->2018) =", round((gli_med.iloc[-1] - gli_med.iloc[0]) / (len(gli_med) - 1), 2), "lakh")

print("T4g: 5 most expensive ads")
print(df.nlargest(5, "Price")[["Brand", "Model", "Year", "KMs Driven", "Price_lakh", "Registered City"]])
print("T4h: Transaction Type")
print(df.groupby("Transaction Type")["Price_lakh"].agg(["count", "median"]))
```

**T4f answer:** A Corolla GLI gets cheaper as it gets older: the median price rises from 12.25 lakh (2010) to 23.2 lakh (2018). The newer model is on average about **1.37 lakh** more expensive per year of difference, so an older car loses roughly 1.4 lakh of value for every extra year of age (the gap is bigger for the newest years, 2 to 2.2 lakh).

**T4g answer:** Yes. The **2015 Toyota Prius at 357.5 lakh** (Rs 35.75 million) looks wrong. A 2015 Prius normally sells for roughly 30 to 40 lakh, so it is probably a decimal or typing mistake (35.75 lakh written as 357.5). The 2007 Surf at 410 lakh with only 3,222 km also looks suspicious. The Land Cruisers (670, 520 and 420 lakh) are possible outliers because new Land Cruisers are very expensive, so they should be checked, not deleted automatically.

**T4h answer:** Cash ads are the overwhelming majority (19,634, median 6.9 lakh) and Installment/Leasing ads are few (850, median 3.0 lakh). Installment cars are less than half as expensive, which suggests they are mostly cheaper, older cars offered by dealers or sellers who accept instalments, while expensive cars are sold for cash.
